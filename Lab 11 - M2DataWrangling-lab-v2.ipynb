### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# Check unique values in Country
print("Unique countries:", df['Country'].nunique())
print("\nTop 20 countries:")
print(df['Country'].value_counts().head(20))

# Look for potential inconsistencies (example: very rare or unusual entries)
print("\nLeast common countries (possible inconsistencies):")
print(df['Country'].value_counts().tail(10))
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
edlevel_mapping = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor's",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master's",
    "Some college/university study without earning a degree": "Some College",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)": "Secondary",
    "Associate degree (A.A., A.S., etc.)": "Associate",
    "Professional degree (JD, MD, Ph.D, Ed.D, etc.)": "Professional",
    "Primary/elementary school": "Primary",
    "Something else": "Other"
}

df['EdLevel'] = df['EdLevel'].replace(edlevel_mapping)

print("Standardized EdLevel value counts:")
print(df['EdLevel'].value_counts())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
# One-hot encode the Employment column
employment_dummies = pd.get_dummies(df['Employment'], prefix='Employment')

# Concatenate with original dataframe
df = pd.concat([df, employment_dummies], axis=1)

print("One-hot encoded Employment columns:")
print(employment_dummies.columns.tolist())
print("\nFirst few rows of encoded columns:")
print(employment_dummies.head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
missing_counts = df.isnull().sum().sort_values(ascending=False)
print("Top 15 columns with most missing values:")
print(missing_counts.head(15))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
# Use median (more robust to outliers for compensation data)
median_comp = df['ConvertedCompYearly'].median()
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(median_comp)

print("Missing values in ConvertedCompYearly after imputation:", df['ConvertedCompYearly'].isnull().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# Most frequent value
majority_remote = df['RemoteWork'].mode()[0]
df['RemoteWork'] = df['RemoteWork'].fillna(majority_remote)

print("Most frequent RemoteWork value:", majority_remote)
print("Missing values in RemoteWork after imputation:", df['RemoteWork'].isnull().sum())
```

<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
# First, convert YearsCodePro to numeric (it often contains strings like "Less than 1 year")
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# Create ExperienceLevel bins
def experience_level(years):
    if pd.isna(years):
        return 'Unknown'
    elif years < 3:
        return 'Junior'
    elif years < 7:
        return 'Mid-level'
    elif years < 12:
        return 'Senior'
    else:
        return 'Expert'

df['ExperienceLevel'] = df['YearsCodePro'].apply(experience_level)

print("ExperienceLevel distribution:")
print(df['ExperienceLevel'].value_counts())
```
