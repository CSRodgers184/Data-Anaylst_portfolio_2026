### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
# Number of duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())

# Optional: show the actual duplicate rows
df[df.duplicated()]
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python

# Remove duplicates
df = df.drop_duplicates()

# Verify they are gone
print("Number of duplicate rows after removal:", df.duplicated().sum())
print("New shape of the dataframe:", df.shape)## Write your code here
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
# Count of missing values in every column
missing_values = df.isnull().sum()
print(missing_values)

# Optional: show only columns that have missing values
print("\nColumns with missing values:")
print(missing_values[missing_values > 0])
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
print("Number of missing values in 'RemoteWork':")
print(df['RemoteWork'].isnull().sum())
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
print(df['RemoteWork'].value_counts(dropna=False))
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
majority = df['RemoteWork'].mode()[0]
print("Most frequent (majority) value in RemoteWork:", majority)
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
# Impute missing values
df['RemoteWork'] = df['RemoteWork'].fillna(majority)

# Verify
print("Missing values in 'RemoteWork' after imputation:")
print(df['RemoteWork'].isnull().sum())
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
# Find compensation-related columns
comp_cols = [col for col in df.columns if 'Comp' in col or 'Salary' in col or 'Pay' in col]
print("Compensation-related columns found:", comp_cols)

# Describe their distribution
for col in comp_cols:
    print(f"\n=== {col} ===")
    print(df[col].describe())
    print(f"Missing values: {df[col].isnull().sum()}")
```
